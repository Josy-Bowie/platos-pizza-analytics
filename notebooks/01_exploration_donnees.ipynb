
### 4.3 Contrôle de l'unicité des identifiants

**Objectif :**

Vérifier que chaque table possède un identifiant unique pour chaque enregistrement.

**Méthodologie :**

- Comparer le nombre total de lignes avec le nombre d'identifiants distincts.
- Utiliser `nunique()` pour compter les valeurs uniques.
- Utiliser `duplicated(subset=[...])` pour détecter les identifiants répétés.
- Identifier les anomalies avant la création des clés primaires SQL.

**Enjeu métier :**

Garantir la fiabilité des relations entre les tables et éviter les erreurs de calcul lors des jointures SQL.


```python
# 4.3 Contrôle de l'unicité des identifiants
# Table ORDERS
print("UNICITE - ORDERS")
print("Nombre de ligne:", len(orders))
print("Identifiants unique:", orders["order_id"].nunique())
print("Identifiants dupliqués:", orders.duplicated(subset=["order_id"]).sum())

# Table ORDER_DETAILS
print("\nUNICITÉ - ORDER_DETAILS")
print("Nombre de lignes :", len(order_details))
print("Identifiants uniques :", order_details["order_details_id"].nunique())
print("Identifiants dupliqués :", order_details.duplicated(subset=["order_details_id"]).sum())

# Table PIZZAS
print("\nUNICITÉ - PIZZAS")
print("Nombre de lignes :", len(pizzas))
print("Identifiants uniques :", pizzas["pizza_id"].nunique())
print("Identifiants dupliqués :", pizzas.duplicated(subset=["pizza_id"]).sum())

# Table PIZZA_TYPES
print("\nUNICITÉ - PIZZA_TYPES")
print("Nombre de lignes :", len(pizza_types))
print("Identifiants uniques :", pizza_types["pizza_type_id"].nunique())
print("Identifiants dupliqués :", pizza_types.duplicated(subset=["pizza_type_id"]).sum())
```


### 4.4 Contrôle de l'intégrité référentielle

**Objectif :**

Vérifier que les clés étrangères des tables référencent des identifiants existants.

**Relations contrôlées :**

1. `order_details.order_id` → `orders.order_id`
2. `order_details.pizza_id` → `pizzas.pizza_id`
3. `pizzas.pizza_type_id` → `pizza_types.pizza_type_id`

**Méthodologie :**

Utiliser la méthode Pandas `isin()` pour vérifier l'existence des identifiants dans les tables parentes.

**Enjeu métier :**

Éviter les lignes orphelines, garantir la cohérence des jointures SQL et fiabiliser les KPI commerciaux.


```python

# 4.4 Contrôle de l'intégrité référentielle

# 1. Vérifier les commandes référencées
commandes_orphelines = ~order_details["order_id"].isin(
    orders["order_id"]
)

print("COMMANDES SANS RÉFÉRENCE :")
print(commandes_orphelines.sum())

# 2. Vérifier les pizzas référencées
pizzas_orphelines = ~order_details["pizza_id"].isin(
    pizzas["pizza_id"]
)

print("\nPIZZAS SANS RÉFÉRENCE :")
print(pizzas_orphelines.sum())

# 3. Vérifier les types de pizzas référencés
types_orphelins = ~pizzas["pizza_type_id"].isin(
    pizza_types["pizza_type_id"]
)

print("\nTYPES DE PIZZAS SANS RÉFÉRENCE :")
print(types_orphelins.sum())
```


**Résultats et interprétation :**

Les trois contrôles d'intégrité référentielle n'ont détecté aucune référence orpheline.

- Toutes les lignes de commande référencent une commande existante.
- Toutes les pizzas commandées référencent une pizza existante.
- Toutes les pizzas du catalogue référencent un type de pizza existant.

**Conclusion métier :**

Les relations entre les quatre tables sont cohérentes.Les données peuvent être utilisées pour préparer la modélisation relationnelle SQL.

Les autres contrôles de qualité restent à effectuer.



### 4.5 Contrôle de la cohérence métier

**Objectif :**

Identifier les valeurs numériques incohérentes susceptibles de fausser les indicateurs commerciaux.

**Contrôles :**

1. Vérifier que les quantités vendues sont strictement positives.
2. Vérifier que les prix des pizzas sont strictement positifs.
3. Examiner les valeurs minimales et maximales.

**Enjeu métier :**

Garantir la fiabilité des calculs du chiffre d'affaires et des quantités vendues.


```python

# 4.5 Contrôle de la cohérence métier

# 1. Examiner les quantités vendues
print("CONTRÔLE DES QUANTITÉS")

print("Quantité minimale :", order_details["quantity"].min())
print("Quantité maximale :", order_details["quantity"].max())

print(
    "Quantités nulles ou négatives :",
    (order_details["quantity"] <= 0).sum()
)

# 2. Examiner les prix des pizzas
print("\nCONTRÔLE DES PRIX")

print("Prix minimum :", pizzas["price"].min())
print("Prix maximum :", pizzas["price"].max())

print(
    "Prix nuls ou négatifs :",
    (pizzas["price"] <= 0).sum()
)
```


**Résultats du contrôle de cohérence métier**

- Quantité minimale : 1
- Quantité maximale : 4
- Quantités nulles ou négatives : 0
- Prix minimum : 9,75
- Prix maximum : 35,95
- Prix nuls ou négatifs : 0

**Interprétation métier :**

Les quantités vendues et les prix des pizzas respectent les règles de positivité définies.

Aucune anomalie numérique n'a été détectée sur les quantités et les prix.

Ces résultats constituent un premier niveau de fiabilisation des futurs calculs du chiffre d'affaires.



### 4.6 Vérification de la cohérence des dates

**Objectif :**

Vérifier la validité des dates et des heures des commandes avant les analyses temporelles.

**Contrôles réalisés :**

1. Convertir les dates et heures au format datetime.
2. Identifier les valeurs impossibles à convertir.
3. Déterminer les dates minimale et maximale.
4. Vérifier la période couverte par les données.

**Enjeu métier :**

Garantir la fiabilité des analyses de ventes par jour, par mois et par heure.


```python

# 4.6 Vérification des dates des commandes

# 1. Convertir les dates en datetime
dates_converties = pd.to_datetime(
    orders["date"],
    format="%Y-%m-%d",
    errors="coerce"
)

# 2. Convertir les heures
heures_converties = pd.to_datetime(
    orders["time"],
    format="%H:%M:%S",
    errors="coerce"
)

# 3. Vérifier les conversions
print("CONTRÔLE DES DATES")

print("Dates invalides :", dates_converties.isna().sum())
print("Heures invalides :", heures_converties.isna().sum())

# 4. Identifier la période
print("\nPÉRIODE DES COMMANDES")

print("Première date :", dates_converties.min())
print("Dernière date :", dates_converties.max())

# 5. Compter les jours distincts
print(
    "Nombre de jours distincts :",
    dates_converties.nunique()
)
```
